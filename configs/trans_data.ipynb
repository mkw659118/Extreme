```python
import os
import scipy.io
import numpy as np

mat_path = "/home/xuke/mkw/Extreme/datasets/Seattle/Seattle.mat"
save_path = "/home/xuke/mkw/Extreme/datasets/Seattle/Seattle.csv"

mat = scipy.io.loadmat(mat_path)
data = mat["data"]

# Seattle: [T, N, N] -> [T, N*N]
data_2d = data.reshape(data.shape[0], -1)

np.savetxt(save_path, data_2d, delimiter=",", fmt="%.8g")

print("original shape:", data.shape)
print("csv shape:", data_2d.shape)
print("saved to:", save_path)
```
