# Abilene Dataset Visualization

This notebook loads and visualizes `Abilene_12_12_3000_T3000_flat144.csv`.

## 加载数据集与环境依赖

Load the CSV with pandas, import plotting libraries, and show shape/columns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

csv_path = "datasets/Abilene_12_12_3000_T3000_flat144.csv"

# The file has no header, so create generic column names.
df = pd.read_csv(csv_path, header=None)
df.columns = [f"c{i}" for i in range(df.shape[1])]

print("shape:", df.shape)
print("columns:", df.columns.tolist()[:10], "...", df.columns.tolist()[-3:])
```

## 缺失值与列类型检查

Count missing values, check dtypes, and coerce to numeric if needed.

```python
missing = df.isna().sum()
missing_rate = (missing / len(df)).sort_values(ascending=False)
print("missing count (top 10):\n", missing.sort_values(ascending=False).head(10))
print("missing rate (top 10):\n", missing_rate.head(10))
print("dtypes:", df.dtypes.unique())

# Ensure numeric types
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

## 可视化：数值分布（直方图/箱线图）

Plot histograms and boxplots for a subset of variables.

```python
cols = df.columns[:6]
fig, axes = plt.subplots(2, 3, figsize=(12, 6))
axes = axes.flatten()

for ax, col in zip(axes, cols):
    ax.hist(df[col].values, bins=50, color="#3b82f6", alpha=0.75)
    ax.set_title(col)
    ax.set_xlabel("value")
    ax.set_ylabel("count")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df[cols], orient="h")
plt.title("Boxplot of selected variables")
plt.show()
```

## 可视化：相关性热力图

Compute correlation for a subset of columns to keep the plot readable.

```python
subset_cols = df.columns[:20]
plt.figure(figsize=(8, 6))
sns.heatmap(df[subset_cols].corr(), cmap="coolwarm", center=0)
plt.title("Correlation heatmap (first 20 variables)")
plt.show()
```

## 可视化：时间/索引趋势图

Plot time/index trends for selected variables.

```python
plt.figure(figsize=(12, 4))
for col in df.columns[:5]:
    plt.plot(df.index, df[col], label=col, linewidth=1)
plt.legend(ncol=5, fontsize=8)
plt.title("Time/index trends (first 5 variables)")
plt.xlabel("time index")
plt.ylabel("value")
plt.show()
```
