```python
import os
import numpy as np

src_path = "./datasets/Seattle/Seattle.csv"
save_path = "./datasets/Seattle/Seattle_single.csv"

target_col = 1

data = np.loadtxt(src_path, delimiter=",")

if data.ndim == 1:
    data = data.reshape(-1, 1)

target_series = data[:, target_col:target_col + 1]

os.makedirs(os.path.dirname(save_path), exist_ok=True)
np.savetxt(save_path, target_series, delimiter=",", fmt="%.8g")

print("original shape:", data.shape)
print("single-variable shape:", target_series.shape)
print("saved to:", save_path)
```
